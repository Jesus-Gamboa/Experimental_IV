#**Determinación del coeficiente de expansión**

El coeficiente de expansión de define como:

$$α=\frac{ΔL}{LΔT}\,\,(1)$$

Donde:
$α:$ Coeficiente de expansión térmico lineal.

$L:$ Longitud inicial de la barra.

$\Delta L:$ Variación de la longitud.

$\Delta T:$ Variación de la temperatura

Y su incertidumbre puede determinarse como:

$$ δα=\sqrt{(\frac{1}{LΔT})^2(δΔL)^2 + (\frac{ΔL}{ΔT}In(L))^2(δL)^2 + (\frac{ΔL}{L}In(\Delta T))^2(δΔT)^2} \,(2)$$

Donde:

$δΔL:$ Incertidumbre de la variación de longitud de la barra.

$δL:$: Incertidumbre de la longitud inicial de la barra.

$δΔ T:$ Incertidumbre de la variación de la temperatura

Adicionalmente, se puede calcular la incertidumbre de la temperatura promedio como:

$$\delta T_{prom}= \frac{1}{2}\sqrt{(\delta T_f)^2 + (\delta T_i)^2} \,\,(3)$$

Finalmente, la incertidumbre de la suma o resta de temperaturas se determina como:

$$\delta \Delta T = \sqrt{(\delta T_f)^2 + (\delta T_i)^2} \,\,(4)$$

$$ϵ_{\%}=\frac{|V_{aceptado}- V_{experimental}|}{V_{aceptado}}× 100 \,\,\,(5)$$

```python
def coeficiente_expansion (delta_l, l, T_inicial, T_final):
  """
  Determina el coeficiente de expansión empleando la ecuación (1).
  Entradas:
    delta_l: variación de la longitud.
    l: longitud inicial.
    T_inicial: temperatura inicial.
    T_final: temperatura final.
  Salidas:
    alpha: Coeficiente de expansión.
  """
  alpha=delta_l/((l)*(T_final-T_inicial))
  return alpha
```

```python
def incertidumbre_delta_T(delta_Ti,delta_Tf):
  """
  Determina la incertidumbre de la variación de la temperatura con (4).
  Entradas:
    delta_Ti: Incertidumbre de la temperatura inicial.
    deta_Tf: Incertidumbre de la temperatura final.
  Salidas:
    delta_T: incertidumbre de la variación de temperatura.
  """
  delta_T=np.sqrt((delta_Tf**2) + (delta_Ti**2))
  return delta_T
```

```python
def incertidumbre_alpha (delta_l, l, T_inicial, T_final, inc_delta_l, inc_l, inc_delta_T):
  """
  Determina la incertidumbre del coeficiente de expansión usando (2).
  Entradas:
    delta_l: variación de la longitud.
    l: longitud inicial.
    T_inicial: temperatura inicial.
    T_final: temperatura final.
    inc_delta_l: Incertidumbre de la variación de la longitud.
    inc_l: Incertidumbre de la variación de l.
    inc_delta_T: Incertidumbre de la variación de la temperatura.
  Salidas:
    inc_alpha: Incertidumbre del coeficiente de expansión.
  """
  inc_alpha = np.sqrt(
    (1/(l*(T_final-T_inicial)))**2 * (inc_delta_l)**2 +
    ((delta_l/(T_final-T_inicial)) * (1/l**2))**2 * (inc_l**2) +
    ((delta_l/l) * (1/(T_final-T_inicial)**2))**2 * (inc_delta_T)**2)
  return inc_alpha
```

```python
def incertidubre_temperatura_promedio (delta_Ti,delta_Tf):
  """
  Determina la incertidumbre de la variación de la temperatura con (3).
  Entradas:
    delta_Ti: Incertidumbre de la temperatura inicial.
    deta_Tf: Incertidumbre de la temperatura final.
  Salidas:
    delta_T_prom: incertidumbre de la variación de temperatura promedio.
  """
  delta_T_prom=np.sqrt((delta_Tf**2) + (delta_Ti**2))/2
  return delta_T_prom
```

```python
def error_porcentual (valor_aceptado, valor_experimental):
  """
  Determina el error porcentual por medio de la ecuación 5
  Entradas:
    valor_aceptado: Valor aceptado de una variable
    valor_experimental: valor de la variable obtenida experimentalmente
  Salidas:
    epsilon: error porcentual
  """
  epsilon= (abs(valor_aceptado-valor_experimental)/valor_aceptado)*100
  return epsilon
```
